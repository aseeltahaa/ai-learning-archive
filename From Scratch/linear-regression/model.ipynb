```python
import numpy as np
```

```python
class LinearRegression:
    """
    Attributes:
    learning_rate: Step size used during gradient descent updates.
    number_of_iterations : Number of iterations for the gradient descent optimization.
    w : Weight vector for the features.
    b : Bias (intercept) term.
    m : Number of training samples.
    n : Number of features.
    """

    # Constructor: initializes the hyperparameters of the model
    def __init__(self, learning_rate, number_of_iterations):
        self.learning_rate = learning_rate
        self.number_of_iterations = number_of_iterations
       
        
    # Trains the Linear Regression model using Gradient Descent
    def fit(self, X, Y):
        # Number of training examples (rows) and number of features (columns)
        self.m, self.n = X.shape

        # Initialize weights as zeros and bias as 0
        self.w = np.zeros(self.n)
        self.b = 0

        # Store training data for gradient calculations
        self.X = X
        self.Y = Y

        # Perform Gradient Descent optimization
        for _ in range(self.number_of_iterations):
            self.update_weights()
      
            
    # Updates the weights and bias using Gradient Descent
    def update_weights(self):
        # Predict values using the current model parameters
        Y_prediction = self.predict(self.X)

        # Compute gradients of the Mean Squared Error loss function
        dw = -(2 / self.m) * self.X.T.dot(self.Y - Y_prediction)
        db = -(2 / self.m) * np.sum(self.Y - Y_prediction)

        # Update weights and bias using gradient descent rule
        self.w = self.w - self.learning_rate * dw
        self.b = self.b - self.learning_rate * db
    
    
    # Predicts target values using the trained Linear Regression model
    def predict(self, X):
        # Linear regression equation: y = Xw + b
        return X.dot(self.w) + self.b
```
