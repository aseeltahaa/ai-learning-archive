```python
import matplotlib.pyplot as plt
import numpy as np
'''
  A python library for data visualization
'''
```

```python
x = np.array([2023, 2024, 2025, 2026])
y1 = np.array([15, 25, 30, 20])
y2 = np.array([20, 30, 25, 10])

line_style = dict(
    marker = ".",
    markersize = 20, 
    markerfacecolor = "white", 
    markeredgecolor = "black",
    linestyle = "solid",
    linewidth = 2
)

plt.title("Class size", fontsize = 20, family = "Arial", fontweight = "bold", color = "grey")

plt.xlabel("Year", fontsize = 20, family = "Arial")
plt.ylabel("Nb of Students",  fontsize = 20, family = "Arial" )

plt.plot(x, y1, **line_style, color = "red")
plt.plot(x, y2, **line_style, color = "blue")

plt.xticks(x)
plt.tick_params(axis = "both", colors = "blue")

plt.show()
```

```python
plt.plot(y1)
plt.plot(y2)
plt.show()
```

Customize Plots

```python
plt.plot(x, y1, marker = ".",
          markersize = 20, 
          markerfacecolor = "cyan", 
          markeredgecolor = "blue",
          linestyle = "dotted",
          linewidth = 3, 
          color = "blue")
plt.show()
```

Grid Lines

```python
x = [1, 2, 3, 4, 5]
y = [5,10,15, 20, 25]

plt.grid(axis = "x", linewidth = 1.5, color = "lightgrey", linestyle ="dotted")

plt.plot(x, y)
plt.show()
```

**Bar Charts**  are used to compare categories of data by representing each category with a bar

```python
# Food Pyramid: Daily Comparisons
categories = np.array(["Grains", "Fruit", "Vegetables", "Protein", "Dairy", "Sweets"])
values = np.array([4, 3, 2, 5, 3, 1])

plt.title("Daily Consumption")
plt.xlabel("Food")
plt.ylabel("Quantity")

plt.bar(categories, values, color = "cyan")
plt.show()
```

```python
plt.title("Daily Consumption")
plt.xlabel("Quantity")
plt.ylabel("Food")

plt.barh(categories, values, color = "cyan")
plt.show()
```

**Pie Charts** are circular charts divided into slices to show percentages of the total.

```python
categories = ["Freshmen", "Sophmores", "Juniors", "Seniors"]
values = np.array([300, 250, 275, 225])
colors = ["red", "yellow", "blue", "green"]

plt.title = ("College Fake Data")
plt.pie(values, 
        labels = categories, 
        autopct = "%1.1f", 
        colors = colors,
        explode = [0,0,0,0.2],
        shadow = True,
        startangle = 90)
plt.show()
```

**Scatter Graphs** show the relationship between two variables; it helps us identify the correlation between them (+, -, None)

```python
x1 = np.array([0, 1 , 1, 2, 3, 4, 5, 6, 7, 7, 8])
y1 = np.array([55, 60, 65, 62, 68, 70, 75, 78, 82, 85, 87])

x2 = np.array([0, 1, 2, 2, 3, 4, 5, 5, 7, 8, 8])
y2 = np.array([50, 58, 65, 70, 72, 78, 83, 88, 92, 95, 97])


plt.xlabel("Studied hours")
plt.ylabel("Grade")

plt.scatter(x1, y1, color = "red", alpha = 0.7, s = 150, label = "Class 1")
plt.scatter(x2, y2, color = "blue", alpha = 0.7, s = 150, label = "Class 2")

plt.legend()
plt.show()
```

**Histograms** represent the distribution of quantitative data by grouping values into bins and counting their frequencies

```python
scores = np.random.normal(loc=80, scale=10, size=100)
scores = np.clip(scores, 0, 100)

plt.hist(scores, bins=10, color="lightgreen", edgecolor="green")

plt.xlabel("Score")
plt.ylabel("Number of students")

plt.show()
```

**Subplots** returns a tuple object that gets unpacked

```python
figure, axes = plt.subplots(2, 2)
x = np.array([1,2,3,4,5])

axes[0, 0].plot(x, x*2, color = "red")
axes[0,0].set_title("x*2")

axes[0, 1].plot(x, x**2, color = "blue")
axes[0,1].set_title("x ^ 2")

axes[1, 0].plot(x, x**3, color = "yellow")
axes[1,0].set_title("x ^ 3")

axes[1, 1].plot(x, x**4, color = "green")
axes[1, 1].set_title("x ^ 4")

plt.tight_layout()
plt.show()
```
