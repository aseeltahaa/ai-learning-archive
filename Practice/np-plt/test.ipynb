```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0, 10, 100)
y = np.sin(x)
z = np.cos(x)
x
```

```python
plt.plot(x, y)
plt.xlabel('x')
plt.ylabel('sin(x)')
plt.title('Sine Wave')
plt.show()
```

```python
plt.plot(x, z)
plt.xlabel('x')
plt.ylabel('cos(x)')
plt.title('Cosine Wave')
plt.show()
```

```python
# parabola
x = np.linspace(-10,10,20)
y = x**2
plt.plot(x,y)
plt.show()
```

```python
plt.plot(x, y, 'r+')
plt.show()
```

```python
plt.plot(x, y, 'gx')
plt.show()
```

```python
x = np.linspace(-5,5,50)
plt.plot(x, np.sin(x), 'g-')
plt.plot(x, np.cos(x), 'r--')
plt.show()
```

```python
fig = plt.figure()
languages = ['English','French','Spanish','Latin','German']
people = [100, 50, 150, 40, 70]
plt.bar(languages, people)
plt.xlabel('LANGUAGES')
plt.ylabel('NUMBER OF PEOPLE')
plt.show()
```

```python
languages = ['English','French','Spanish','Latin','German']
people = [100, 50, 150, 40, 70]
plt.pie(people, labels=languages, autopct='%1.1f%%')
plt.show()
```

```python
x = np.linspace(0,10,30)
y = np.sin(x)
z = np.cos(x)
plt.scatter(x,y,color='g')
plt.scatter(x,z,color='b')
plt.show()
```
