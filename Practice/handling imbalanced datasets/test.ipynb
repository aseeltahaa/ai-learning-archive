```python
import numpy as np
import pandas as pd
```

```python
credit_card_ds = pd.read_csv('credit_data.csv')
credit_card_ds.head()
```

```python
# No missing values in this ds
credit_card_ds.isnull().sum()
```

```python
# imbalanced data
credit_card_ds['Class'].value_counts()
```

0 -> Legit Transactions
1 -> Fraudulant Transactions

```python
# separating the data
legit = credit_card_ds[credit_card_ds.Class == 0]
fraud = credit_card_ds[credit_card_ds.Class == 1]
print(legit.shape)
print(fraud.shape)
```

```python
legit_sample = legit.sample(n = 492)
print(legit_sample.shape)
```

```python
new_df = pd.concat([legit_sample, fraud], axis = 0)
new_df.head()
```
