### Import Data

```python
crash_df = sns.load_dataset('car_crashes')
crash_df.head()
```

#### Distribution Plots

Distribution Plot

```python
sns.displot(crash_df['not_distracted'], kde = True, bins = 20)
```

Joint Plot

```python
sns.jointplot(x = 'speeding', y = 'alcohol', data = crash_df, kind = 'reg')
```

KDE

```python
sns.jointplot(x = 'speeding', y = 'alcohol', data = crash_df, kind = 'kde')
```

Hexagon Plot

```python
sns.jointplot(x = 'speeding', y = 'alcohol', data = crash_df, kind = 'hex')
```

Pair Plots

```python
sns.pairplot(crash_df)
```

```python
tips_df = sns.load_dataset('tips')
sns.pairplot(tips_df, hue = 'sex', palette='Reds')
```

Rug Plot

```python
sns.rugplot(tips_df['tip'])
```

#### Styling

```python
sns.set_style('darkgrid')
plt.figure(figsize = (8, 4))
sns.set_context('paper', font_scale=1.2)
sns.jointplot(x = 'speeding', y = 'alcohol', data = crash_df, kind = 'reg')
```

#### Bar Plot

```python
sns.barplot(x = 'sex', y = 'total_bill', data = tips_df, estimator= np.median)
```

#### Box Plot

```python
sns.boxplot(x = 'day', y = 'total_bill', data = tips_df, hue = 'sex')
```

#### Violin Plot

```python
sns.violinplot(x = 'day', y = 'total_bill', data = tips_df, hue = 'sex', split = True)
```

#### Strip Plots

```python
sns.stripplot(x = 'day', y = 'total_bill', data = tips_df, jitter = True, hue = 'sex', dodge= True)
```

Swarm Plot

```python
sns.swarmplot(x = 'day', y = 'total_bill', data = tips_df, color = "darkblue")
sns.violinplot(x = 'day', y = 'total_bill', data = tips_df)
```

#### Heatmaps

```python
crash_mx = crash_df.select_dtypes(include='number').corr()
sns.heatmap(crash_mx, annot = True, cmap = 'Blues')
```
