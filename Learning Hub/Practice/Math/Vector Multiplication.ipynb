```python
import numpy as np
```

# Dot Product
It helps us determine how aligned two vectors are: 

If result > 0 -> same direction-ish 

If the result = 0 -> perpendicular

If result < 0 -> opposite-ish

```python
a = np.array([2, 3])
b = np.array([4, 4])
a_dot_b = np.dot(a, b)
a_dot_b
```

```python
b_dot_a = np.dot(b, a)
b_dot_a
```

# Cross Product 
Result is a vector perpendicular to both

```python
a = np.array([2, 3])
b = np.array([4, 4])
a_cross_b = np.cross(a, b)
a_cross_b
```

```python
c = np.array([5, 10, 20])
d = np.array([18, 32, 50])
c_cross_d = np.cross(c, d)
c_cross_d
```
