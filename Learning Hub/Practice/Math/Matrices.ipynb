# Matrix Operations

```python
A = np.array([[2,3], [4,5]])
A
```

```python
B = np.array([[6,7], [8,9]])
B
```

## Adding 2 Matrices

```python
sum = A + B
sum
```

```python
sum2 = np.add(A, B)
sum2
```

## Matrix Subtraction

```python
sub = A - B
sub
```

```python
sub2 = np.subtract(A, B)
sub2
```

## Multiplying by a Scalar

```python
sc = 5 * A
sc
```

```python
sc2 = np.multiply(5, A)
sc2
```

## Multiply 2 Matrices

```python
prod = A * B 
prod
```

```python
prod2 = np.multiply(A, B)
prod2
```
