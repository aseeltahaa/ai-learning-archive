Import the dependencies

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from xgboost import XGBRegressor
from sklearn import metrics
```

Import Dataset

```python
house_price_dataset = fetch_california_housing()
house_price_dataset
```

```python
#  Loading the dataset into a Pandas dataframe
house_price_dataframe = pd.DataFrame(
    house_price_dataset.data, 
    columns=house_price_dataset.feature_names
)
house_price_dataframe.head()
```

```python
# add the price column to the dataframe
house_price_dataframe['price'] = house_price_dataset.target
house_price_dataframe.head()
```

```python
# number of rows and columns
house_price_dataframe.shape
```

Preprocessing

```python
# Check for missing values
house_price_dataframe.isnull().sum()
```

```python
# statistical measures
house_price_dataframe.describe()
```

Understanding Feature Corellations:

```python
correlation = house_price_dataframe.corr()
```

```python
# construct heatmap
plt.figure(
    figsize = (10, 10)
)
sns.heatmap(
    correlation, 
    cbar = True, 
    fmt = '.1f', 
    annot = True, 
    annot_kws = {'size': 8}, 
    cmap = 'Blues'
)
```

Train Test Split

```python
X = house_price_dataframe.drop(['price'], axis = 1)
Y = house_price_dataframe['price']
```

```python
X
```

```python
Y
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state = 2)
```

```python
X.shape, X_train.shape, X_test.shape
```

Model Training: XGBoost Regressor

```python
# loading the model
model = XGBRegressor()
# training the model
model.fit(X_train, Y_train)
```

Model Evaluation

```python
# Prediction on training data
train_data_prediction = model.predict(X_train)
score1 = metrics.r2_score(Y_train, train_data_prediction)
score1
```

```python
score2 = metrics.mean_absolute_error(Y_train, train_data_prediction)
score2
```

```python
# Prediction on testing data
test_data_prediction = model.predict(X_test)
score1 = metrics.r2_score(Y_test, test_data_prediction)
score1
```

```python
score2 = metrics.mean_absolute_error(Y_test, test_data_prediction)
score2
```
